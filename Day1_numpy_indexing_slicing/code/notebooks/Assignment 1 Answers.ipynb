```python
3D Array

Section A – Access Individual Elements

print(arr[0,0,0])   # 10
print(arr[0,2,4])   # 24
print(arr[1,1,2])   # 37
print(arr[1,2,3])   # 43
print(arr[2,0,4])   # 54
print(arr[2,3,1])   # 66
print(arr[2,1,3])   # 58
print(arr[1,3,4])   # 49

Section B – Access Complete Rows
print(arr[0,1,:])
print(arr[1,2,:])
print(arr[2,3,:])

Section C – Access Complete 2D Matrices
print(arr[0])
print(arr[1])
print(arr[2])

Section D – Column Extraction
print(arr[0,:,0])
print(arr[1,:,2])
print(arr[2,:,4])

Section E – Slicing
# First two matrices
print(arr[:2])

# Last two matrices
print(arr[1:])

# First two rows from every matrix
print(arr[:,:2,:])

# Last two rows from every matrix
print(arr[:,2:,:])

# First three columns from every matrix
print(arr[:,:,:3])

# Last two columns from every matrix
print(arr[:,:,3:])

Section F – Advanced Slicing
print(arr[0,:2,:3])

Section G
# Every alternate matrix
print(arr[::2])

# Every alternate row from all matrices
print(arr[:,::2,:])

# Every alternate column from all matrices
print(arr[:,:,::2])

# First column from every matrix
print(arr[:,:,0])

# Last column from every matrix
print(arr[:,:,-1])

# Middle row from every matrix
print(arr[:,2,:])

# Middle column from every matrix
print(arr[:,:,2])

# Reverse order of matrices
print(arr[::-1])

# Reverse rows inside every matrix
print(arr[:,::-1,:])

# Reverse columns inside every matrix
print(arr[:,:,::-1])
```
