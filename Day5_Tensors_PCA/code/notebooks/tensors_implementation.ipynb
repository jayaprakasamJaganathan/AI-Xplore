## Task 8: Reshape

```python
X = np.arange(24)
print(f"Original X = np.arange(24), shape: {X.shape}, ndim: {X.ndim}")

shapes = [(4, 6), (2, 3, 4), (6, 4), (24,)]

for shape in shapes:
    reshaped = X.reshape(shape)
    print(f"\nReshaped to {shape}:")
    print(f"  Shape: {reshaped.shape}")
    print(f"  ndim: {reshaped.ndim}")
    print(f"  Tensor:\n{reshaped}")

print("\nWhy does reshaping change shape but not total elements?")
print("  Reshaping only changes the view/metadata (strides, shape) of the array.")
print("  The underlying data buffer (memory) remains unchanged.")
print("  Total elements = product of shape dimensions must remain constant.")
print("  This is why reshape requires the new shape to have the same total size.")
```

## Task 9: Reshape vs Transpose

```python
X = np.arange(24).reshape(2, 3, 4)
print(f"Original X (shape {X.shape}):\n{X}")

X_reshaped = X.reshape(3, 2, 4)
print(f"\nX.reshape(3, 2, 4) (shape {X_reshaped.shape}):\n{X_reshaped}")

X_transposed = X.transpose(1, 0, 2)
print(f"\nX.transpose(1, 0, 2) (shape {X_transposed.shape}):\n{X_transposed}")

print(f"\nAre these the same? {np.array_equal(X_reshaped, X_transposed)}")

print("\nExplanation:")
print("  RESHAPE: Reinterprets the memory layout - changes shape but keeps")
print("           the same memory order (C-contiguous by default).")
print("           Elements are read in row-major order and filled into new shape.")
print("  TRANSPOSE: Permutes axes - changes the order of dimensions.")
print("             The data is rearranged so that axis 1 becomes axis 0,")
print("             axis 0 becomes axis 1, axis 2 stays axis 2.")
print("  They are fundamentally different operations!")
```

## Task 10: Broadcasting

```python
X = np.ones((3, 4))
b = np.array([10, 20, 30, 40])

print(f"X.shape = {X.shape}")
print(f"b.shape = {b.shape}")
print(f"X:\n{X}")
print(f"b: {b}")

result_add = X + b
print(f"\nX + b (shape {result_add.shape}):\n{result_add}")

result_mul = X * b
print(f"\nX * b (shape {result_mul.shape}):\n{result_mul}")

print("\nWhy can a (4,) tensor combine with a (3,4) tensor?")
print("  Broadcasting rules (from right to left):")
print("  1. Align shapes on the right: (3, 4) and (4,)")
print("  2. Dimensions are compatible if equal or one is 1")
print("  3. Missing dimensions on the left are treated as 1")
print("  4. b is treated as (1, 4) then broadcast to (3, 4)")
print("  5. The operation is performed element-wise")
```

## Task 11: Broadcasting with Sensor Data

```python
np.random.seed(42)
X = np.random.randn(100, 10) * 5 + 20
offsets = np.array([1.2, 0.8, 1.1, 0.5, 0.9, 1.0, 0.7, 1.3, 0.6, 0.4])

print(f"X.shape = {X.shape} (100 samples × 10 sensors)")
print(f"offsets.shape = {offsets.shape}")
print(f"Offsets: {offsets}")

X_calibrated = X - offsets

print(f"\nX_calibrated.shape = {X_calibrated.shape}")
print(f"First sample before: {X[0]}")
print(f"First sample after:  {X_calibrated[0]}")
print(f"Offsets subtracted:  {X[0] - X_calibrated[0]}")

assert X_calibrated.shape == (100, 10), "Shape should remain (100, 10)"
print("\n✓ Output shape remains (100, 10) as required")
```

## Task 13: Normalize a Tensor

```python
np.random.seed(42)
X = np.random.randn(100, 20, 5) * 10 + 50

print(f"Original X.shape = {X.shape}")
print(f"Original mean per sensor: {X.mean(axis=(0, 1))}")
print(f"Original std per sensor: {X.std(axis=(0, 1))}")

sensor_mean = X.mean(axis=(0, 1), keepdims=True)
sensor_std = X.std(axis=(0, 1), keepdims=True)

print(f"\nsensor_mean shape: {sensor_mean.shape}")
print(f"sensor_std shape: {sensor_std.shape}")

X_normalized = (X - sensor_mean) / sensor_std

print(f"\nNormalized X.shape = {X_normalized.shape}")
print(f"Normalized mean per sensor: {X_normalized.mean(axis=(0, 1))}")
print(f"Normalized std per sensor: {X_normalized.std(axis=(0, 1))}")

assert np.allclose(X_normalized.mean(axis=(0, 1)), 0, atol=1e-10)
assert np.allclose(X_normalized.std(axis=(0, 1)), 1, atol=1e-10)
print("\n✓ Verification passed: mean ≈ 0, std ≈ 1 for each sensor")
```

## Task 14: Flattening

```python
np.random.seed(42)
image = np.random.randint(0, 256, size=(28, 28, 3), dtype=np.uint8)
print(f"Single image shape: {image.shape}")

flattened = image.reshape(-1)
print(f"Flattened shape: {flattened.shape}")
print(f"Expected: (2352,) - 28*28*3 = {28*28*3}")

batch = np.random.randint(0, 256, size=(64, 28, 28, 3), dtype=np.uint8)
print(f"\nBatch shape: {batch.shape}")

batch_flattened = batch.reshape(64, -1)
print(f"Batch flattened shape: {batch_flattened.shape}")
print(f"Expected: (64, 2352)")

print("\nWhy not flatten entire tensor to (150528,)?")
print("  - Would lose the batch structure (which sample is which)")
print("  - Cannot process samples independently in batch operations")
print("  - Loss of sample-level information for training/inference")
print("  - Batch dimension is needed for vectorized operations")
print("  - Model expects (batch_size, features) not (total_elements,)")
```

## Task 15: Concatenation

```python
np.random.seed(42)
visual_features = np.random.randn(32, 128)
sensor_features = np.random.randn(32, 64)

print(f"Visual features shape: {visual_features.shape}")
print(f"Sensor features shape: {sensor_features.shape}")

combined = np.concatenate([visual_features, sensor_features], axis=1)

print(f"Combined shape: {combined.shape}")
print(f"Expected: (32, 192) - 128 + 64 = 192")

print("\nWhy can we concatenate along feature dimension but not arbitrarily?")
print("  - Concatenation requires ALL other dimensions to match exactly")
print("  - Here: batch dimension (32) matches, so we can concat on axis=1")
print("  - Cannot concat on axis=0 because 128 ≠ 64")
print("  - Cannot concat on new axis without reshaping first")
print("  - The concatenation axis is where dimensions ADD; others must be equal")
```
