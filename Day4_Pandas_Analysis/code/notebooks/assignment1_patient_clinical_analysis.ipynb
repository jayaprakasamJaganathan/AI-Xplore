# Assignment 1 — Patient Clinical Data Analysis

Dataset: `patient_clinical_data_raw.csv`

Objective: Clean, transform, analyze, and summarize patient clinical data using Pandas.

```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
```

## Part A — Understanding the Dataset

```python
# 1. Load the dataset into a Pandas DataFrame
df = pd.read_csv('../../data/patient_clinical_data_raw.csv')

# 2. Display the first 10 and last 10 records
print("First 10 records:")
display(df.head(10))

print("\nLast 10 records:")
display(df.tail(10))
```

```python
# 3. Find the number of rows and columns
print(f"Shape: {df.shape}")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
```

```python
# 4. Display the column names and their data types
print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
# 5. Generate a statistical summary of the numerical columns
print("Statistical summary:")
display(df.describe())
```

```python
# 6. Identify the number of unique values in each column
print("Unique values per column:")
for col in df.columns:
    print(f"  {col}: {df[col].nunique()}")
```

## Part B — Data Cleaning

```python
# 7. Identify all columns containing missing values
missing = df.isnull().sum()
missing_cols = missing[missing > 0]
print("Columns with missing values:")
print(missing_cols)
```

```python
# 8. Calculate the percentage of missing values in each column
missing_pct = (missing / len(df)) * 100
print("Percentage of missing values:")
print(missing_pct[missing_pct > 0])
```

```python
# 9. Handle missing values in the numerical columns appropriately
num_cols = df.select_dtypes(include=[np.number]).columns
print("Numerical columns:", num_cols.tolist())

# Fill numerical missing values with median
for col in num_cols:
    if df[col].isnull().any():
        median_val = df[col].median()
        df[col] = df[col].fillna(median_val)
        print(f"Filled {col} missing values with median: {median_val}")
```

```python
# 10. Identify and remove duplicate records
duplicates = df.duplicated().sum()
print(f"Duplicate records: {duplicates}")
if duplicates > 0:
    df = df.drop_duplicates()
    print(f"Removed {duplicates} duplicates. New shape: {df.shape}")
```

```python
# 11. Check the Age column for invalid values
print("Age statistics:")
print(df['Age'].describe())

# Check for negative or unrealistic ages
invalid_age = df[(df['Age'] < 0) | (df['Age'] > 120)]
print(f"\nInvalid age records: {len(invalid_age)}")
if len(invalid_age) > 0:
    display(invalid_age)
```

```python
# 12. Correct or handle patients whose age is outside a reasonable range
# Cap age at reasonable bounds (0-120)
df['Age'] = df['Age'].clip(0, 120)
print("Age clipped to 0-120 range")
print(df['Age'].describe())
```

```python
# 13. Identify inconsistent values in categorical columns such as Gender and Department
cat_cols = df.select_dtypes(include=['object']).columns
print("Categorical columns:", cat_cols.tolist())

for col in cat_cols:
    print(f"\n{col} unique values:")
    print(df[col].value_counts())
```

```python
# 14. Standardize these categorical values
# Example: Standardize Gender
if 'Gender' in df.columns:
    df['Gender'] = df['Gender'].str.strip().str.title()
    print("Gender standardized:")
    print(df['Gender'].value_counts())

# Example: Standardize Department
if 'Department' in df.columns:
    df['Department'] = df['Department'].str.strip().str.title()
    print("\nDepartment standardized:")
    print(df['Department'].value_counts())
```

## Part C — Data Transformation

```python
# 15. Create a new column called Age_Group
def age_group(age):
    if age < 18:
        return 'Child'
    elif age <= 40:
        return 'Young Adult'
    elif age <= 60:
        return 'Middle Age'
    else:
        return 'Senior'

df['Age_Group'] = df['Age'].apply(age_group)
print("Age_Group distribution:")
print(df['Age_Group'].value_counts())
```

```python
# 16. Separate Blood_Pressure into two columns: Systolic and Diastolic
if 'Blood_Pressure' in df.columns:
    bp_split = df['Blood_Pressure'].str.split('/', expand=True)
    df['Systolic'] = pd.to_numeric(bp_split[0], errors='coerce')
    df['Diastolic'] = pd.to_numeric(bp_split[1], errors='coerce')
    print("Blood Pressure split into Systolic and Diastolic")
    print(df[['Blood_Pressure', 'Systolic', 'Diastolic']].head())
else:
    print("Blood_Pressure column not found")
```

```python
# 17. Create a High_Risk column based on suitable clinical conditions
# Example: High risk if Age > 60 OR Systolic > 140 OR Diastolic > 90
conditions = []
if 'Age' in df.columns:
    conditions.append(df['Age'] > 60)
if 'Systolic' in df.columns:
    conditions.append(df['Systolic'] > 140)
if 'Diastolic' in df.columns:
    conditions.append(df['Diastolic'] > 90)

if conditions:
    df['High_Risk'] = np.any(conditions, axis=0)
    print("High_Risk distribution:")
    print(df['High_Risk'].value_counts())
else:
    print("Required columns not available for High_Risk")
```

```python
# 18. Create a Cost_Per_Day column using treatment cost and length of stay
if 'Treatment_Cost' in df.columns and 'Length_of_Stay' in df.columns:
    df['Cost_Per_Day'] = df['Treatment_Cost'] / df['Length_of_Stay'].replace(0, np.nan)
    print("Cost_Per_Day created")
    print(df['Cost_Per_Day'].describe())
else:
    print("Required columns (Treatment_Cost, Length_of_Stay) not found")
```

## Part D — Analysis

```python
# 19. Calculate the average age of patients
print(f"Average age of patients: {df['Age'].mean():.2f}")
```

```python
# 20. Find the number of patients in each department
if 'Department' in df.columns:
    dept_counts = df['Department'].value_counts()
    print("Patients per department:")
    print(dept_counts)
```

```python
# 21. Calculate the average treatment cost for each department
if 'Department' in df.columns and 'Treatment_Cost' in df.columns:
    avg_cost_dept = df.groupby('Department')['Treatment_Cost'].mean().sort_values(ascending=False)
    print("Average treatment cost by department:")
    print(avg_cost_dept)
```

```python
# 22. Find the average length of stay for each diagnosis
if 'Diagnosis' in df.columns and 'Length_of_Stay' in df.columns:
    avg_stay_diag = df.groupby('Diagnosis')['Length_of_Stay'].mean().sort_values(ascending=False)
    print("Average length of stay by diagnosis:")
    print(avg_stay_diag)
```

```python
# 23. Identify the top 10 patients with the highest treatment cost
if 'Treatment_Cost' in df.columns and 'Patient_ID' in df.columns:
    top10_cost = df.nlargest(10, 'Treatment_Cost')[['Patient_ID', 'Treatment_Cost']]
    print("Top 10 patients by treatment cost:")
    print(top10_cost)
```

```python
# 24. Find the department with the highest average treatment cost
if 'Department' in df.columns and 'Treatment_Cost' in df.columns:
    avg_cost_dept = df.groupby('Department')['Treatment_Cost'].mean()
    highest_dept = avg_cost_dept.idxmax()
    highest_cost = avg_cost_dept.max()
    print(f"Department with highest average treatment cost: {highest_dept} ({highest_cost:.2f})")
```

```python
# 25. Calculate the percentage of patients who were readmitted
if 'Readmission' in df.columns:
    readmit_pct = (df['Readmission'] == 'Yes').mean() * 100
    print(f"Readmission rate: {readmit_pct:.2f}%")
```

```python
# 26. Compare readmission rates across departments
if 'Readmission' in df.columns and 'Department' in df.columns:
    readmit_by_dept = df.groupby('Department')['Readmission'].apply(lambda x: (x == 'Yes').mean() * 100).sort_values(ascending=False)
    print("Readmission rate by department:")
    print(readmit_by_dept)
```

```python
# 27. Find the average cholesterol and glucose level by diagnosis
if 'Diagnosis' in df.columns:
    for col in ['Cholesterol', 'Glucose']:
        if col in df.columns:
            avg_by_diag = df.groupby('Diagnosis')[col].mean().sort_values(ascending=False)
            print(f"\nAverage {col} by diagnosis:")
            print(avg_by_diag)
```

```python
# 28. Identify patients with high glucose, high cholesterol, high heart rate
high_risk_patients = df.copy()
conditions = []
if 'Glucose' in df.columns:
    conditions.append(df['Glucose'] > 140)  # High glucose threshold
if 'Cholesterol' in df.columns:
    conditions.append(df['Cholesterol'] > 240)  # High cholesterol threshold
if 'Heart_Rate' in df.columns:
    conditions.append(df['Heart_Rate'] > 100)  # High heart rate threshold

if conditions:
    high_risk = df[np.any(conditions, axis=0)]
    print(f"Patients with high risk indicators: {len(high_risk)}")
    if len(high_risk) > 0:
        display(high_risk[['Patient_ID', 'Glucose', 'Cholesterol', 'Heart_Rate']].head(10))
```
